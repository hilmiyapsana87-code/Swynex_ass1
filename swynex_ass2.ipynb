#Find the most subscribed channel

```python
top_subscribers = df.sort_values(
   by = 'subscribers',
   ascending = False
).head(10)

top_subscribers[['Youtuber','subscribers']]
```

#Visualize the top 10 channels

```python
plt.figure(figsize=(10,6))

sns.barplot(
    data = top_subscribers,
    x = 'subscribers',
    y = 'Youtuber'
)

plt.title('Top 10 Youtube channels by subscribers')
plt.xlabel('Subscribers')
plt.ylabel('Youtuber')

plt.show()
```

#IQR Calculation

```python
Q1 = df["subscribers"].quantile(0.25)
Q3 = df["subscribers"].quantile(0.75)
IQR = Q3 - Q1
```

```python
lower_limit = Q1 - 1.5*IQR
upper_limit = Q3 - 1.5*IQR 
```

#Find outliers

```python
outlier = df[
   (df["subscribers"] < lower_limit)|
   (df["subscribers"] > upper_limit)]
   
```

```python
len(outlier)
```
